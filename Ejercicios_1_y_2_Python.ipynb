Exercise 2
Write a program that:
Asks the user to input two boolean values (True or False).
Evaluates and displays the results of logical operations (and, or, not) between the two values.

```python
frase = input("Ingresa una frase: ")
fraseMayus = frase.upper()
fraseMinus = frase.lower()
titulo = frase.title()

print("Oración en Mayúscula: ", fraseMayus)
print("Oración en minúscula: ", fraseMinus)
print("Oración con primera letra de cada palabra en mayúscula: ", titulo)
```

```python
valor1 = input("Ingresa True o False: ").strip().lower() == "true"
valor2 = input("Ingresa True o False: ").strip().lower() == "true"

print(f"{valor1} AND {valor2} = {valor1 and valor2}")
print(f"{valor1} OR {valor2} = {valor1 or valor2}")
print(f"NOT {valor1} = {not valor1}")
print(f"NOT {valor2} = {not valor2}")
```
