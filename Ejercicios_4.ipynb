Exercise 5
Write a program that:
Uses a for loop to iterate through numbers from 1 to 50.
For multiples of 3, print "Fizz".
For multiples of 5, print "Buzz".
For multiples of both 3 and 5, print "FizzBuzz".
Otherwise, print the number.

```python
num = int(input("Ingresa un número: "))

print(f"Tabla de multiplicar del {num}:")
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

```python
num = int(input("Ingresa un número y adivina el número secreto: "))
numSecreto = 7

while num != numSecreto:
  if num < numSecreto:
    print("Muy bajo")
    num = int(input("Ingresa un número y adivina el número secreto: "))
  elif num > numSecreto:
    print("Muy alto")
    num = int(input("Ingresa un número y adivina el número secreto: "))
print("Correcto!")
```

```python
texto = input("Ingresa una frase o palabra: ").lower()

vocales = "aeiou"

contador = 0

for letra in texto:
    if letra in vocales:
        contador += 1

print(f"La cantidad de vocales en el texto es: {contador}")
```

```python
texto = input("Ingresa una frase o palabra: ")

textoInvertido = ""

for letra in texto:
    textoInvertido = letra + textoInvertido

print(f"Texto invertido: {textoInvertido}")
```

```python
for num in range(1, 51):
    if num % 3 == 0 and num % 5 == 0:
        print("FizzBuzz")
    elif num % 3 == 0:
        print("Fizz")
    elif num % 5 == 0:
        print("Buzz")
    else:
        print(num)
```
