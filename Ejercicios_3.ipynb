```python
anio = int(input("Ingresa un año: "))
if (anio % 4 == 0 and anio % 100 != 0) or (anio % 400 == 0):
  print(f" EL año {anio} es bisiesto")
else:
  print(f"El año {anio} no es bisiesto")
```

```python
numero1 = float(input("Ingresa un numero: "))
numero2 = float(input("Ingresa otro numero: "))
operador = input("ingresa un operador (+,-,* o /)").strip()

match operador:
  case "+":
    print(numero1 + numero2)
  case "-":
    print(numero1 - numero2)
  case "*":
    print(numero1 * numero2)
  case "/":
    if numero2 != 0:
      print(numero1/numero2)
    else:
      print("Error, el divisor no puede ser cero (0)")
  case _:
    print("Otro error")

```

```python
numero = float(input("Ingresa un número: "))

if numero > 0:
    print("El número es positivo.")
elif numero < 0:
    print("El número es negativo.")
else:
    print("El número es cero.")
```

```python
numero = int(input("Ingresa un numero del 1 al 7 (días de la semana): "))


match numero:
  case 1:
    print(f"El número {numero} de la semana es el día Lunes")
  case 2:
    print(f"El número {numero} de la semana es el día Martes")
  case 3:
    print(f"El número {numero} de la semana es el día Miércoles")
  case 4:
    print(f"El número {numero} de la semana es el día Jueves")
  case 5:
    print(f"El número {numero} de la semana es el día Viernes")
  case 6:
    print(f"El número {numero} de la semana es el día Sábado")
  case 7:
    print(f"El número {numero} de la semana es el día Domingo")
  case _:
    print("No existe ese día de la semana")
```
